# 🔀 Routers and Abstraction Layers

## 1. Router

A **router** is a remote middleman that receives our LLM request and **routes it to the model/provider we choose**.

### Flow

Your Application → Router → Selected LLM → Response

### Example: OpenRouter

- OpenRouter acts as a routing layer.
- We can use **one API key** to access many different models/providers.
- We specify which model we want to use.
- OpenRouter sends the request to that model and returns the response.

### Why use a Router?

- One interface/API key for multiple models
- Easy to switch between models
- Avoid managing separate integrations for every provider
- Can compare different models easily

---

## 2. Abstraction Layer

An **abstraction layer** is a piece of code/framework that gives us a **common interface** for working with different LLM providers.

Instead of learning the different API syntax for every provider, the abstraction layer hides those differences.

### Flow

Your Application → Abstraction Layer → OpenAI / Claude / Gemini / etc.

### Main idea

> **Abstraction hides the differences between different APIs.**

---

## 🔥 Router vs Abstraction Layer

| Router | Abstraction Layer |
|---|---|
| Remote service | Usually code/framework on our side |
| Routes requests to different providers | Hides differences between provider APIs |
| Decides **where the request goes** | Gives a **common way to make requests** |
| Example: OpenRouter | Framework/library providing a common interface |

### 🧠 Memory Trick

**Router = Traffic Controller** 🚦  
→ Decides **where the request goes**

**Abstraction Layer = Translator** 🔄  
→ Hides **how each provider works**
