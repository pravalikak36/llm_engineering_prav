# 🔀 Routers and Abstraction Layers

## 1. Router

A **router** is a remote middleman that receives our LLM request and **routes it to the model/provider we choose**.

### Flow

Your Application → Router → Selected LLM → Response

### Example: OpenRouter

- OpenRouter acts as a routing layer.
- We can use **one API key** to access many different models/providers.
- We specify which model we want to use.
- OpenRouter sends the request to that model and returns the response.

### Why use a Router?

- One interface/API key for multiple models
- Easy to switch between models
- Avoid managing separate integrations for every provider
- Can compare different models easily

---

## 2. Abstraction Layer

An **abstraction layer** is a piece of code/framework that gives us a **common interface** for working with different LLM providers.

Instead of learning the different API syntax for every provider, the abstraction layer hides those differences.

### Flow

Your Application → Abstraction Layer → OpenAI / Claude / Gemini / etc.

### Main idea

> **Abstraction hides the differences between different APIs.**

---

## 🔥 Router vs Abstraction Layer

| Router | Abstraction Layer |
|---|---|
| Remote service | Usually code/framework on our side |
| Routes requests to different providers | Hides differences between provider APIs |
| Decides **where the request goes** | Gives a **common way to make requests** |
| Example: OpenRouter | Framework/library providing a common interface |

### 🧠 Memory Trick

**Router = Traffic Controller** 🚦  
→ Decides **where the request goes**

**Abstraction Layer = Translator** 🔄  
→ Hides **how each provider works**

## 🔹 2. LangChain

**LangChain** is a framework for building applications with LLMs.

It provides tools for working with:

- LLMs
- Prompts
- Chains / workflows
- Tools
- Agents
- Retrieval

### Important

```text
LangChain ≠ LLM
LangChain ≠ GPT
```

LangChain is a **framework used to build applications with LLMs**.

### Example

```python
from langchain_openai import ChatOpenAI

llm = ChatOpenAI(model="gpt-5-mini")

response = llm.invoke("Tell me a joke")
```

Instead of directly using the provider's API, LangChain provides its own interface such as:

```python
llm.invoke(...)
```

---

## 🔹 3. LiteLLM

**LiteLLM** is a lightweight abstraction layer that provides a common interface for calling different LLM providers.

### Example

```python
from litellm import completion

response = completion(
    model="openai/gpt-4.1",
    messages=[
        {"role": "user", "content": "Tell me a joke"}
    ]
)
```

The model can be specified using:

```text
provider/model
```

### Main Idea

> Use one common interface to work with many different LLM providers.

---

## 🔹 4. LangChain vs LiteLLM

| LangChain | LiteLLM |
|---|---|
| Full LLM application framework | Lightweight LLM interface |
| Supports chains, tools, agents, retrieval, etc. | Mainly focuses on calling different models consistently |
| More functionality | Simpler and more focused |

### Memory Trick

# 🔹 6. Hallucination

A **hallucination** occurs when an LLM gives an answer that sounds convincing but is incorrect.

### Why?

An LLM generates likely tokens based on the information available in its input.

If it does not have enough information, it may produce a plausible but incorrect answer.

### Important

> **Confident answer ≠ Correct answer**

# 🔹 7. Context

**Context** is the information we provide to the LLM along with the question.

### Example

Instead of asking:

```text
Where is the answer to this question?
```

we provide:

```text
Relevant document
+
Question
```

Then:

```text
Context + Question
        ↓
       LLM
        ↓
    Better answer
```

### Key Idea

> Providing the right context can improve the quality and accuracy of the LLM's response.

# 🔹 8. Prompt Caching

**Prompt caching** allows repeated parts of an input prompt to be reused instead of processing the same content from scratch every time.

### Example

Suppose we repeatedly send a large document:

```text
Large Document
+
Question 1
```

Then:

```text
Large Document
+
Question 2
```

Then:

```text
Large Document
+
Question 3
```

The large repeated part can potentially be cached.

### Flow

```text
First Request
Large Document → Processed + Cached

Later Requests
Large Document → Reused from Cache
Question       → Processed normally
```

### Benefits

- Lower cost for repeated large inputs
- Potentially faster processing
- Useful for large, repeated context

---

# 🔹 9. Cached Tokens

**Cached tokens** are tokens from the input that are reused from a previous request through prompt caching.

```text
Normal input tokens → Normal input cost

Cached input tokens → Usually lower cached-input cost
```

The exact pricing and caching behavior depends on the provider.

---

# 🔹 10. Static vs Dynamic Prompt Content

When using prompt caching, it is useful to separate:

### Static content

Information that does not change frequently.

Examples:

- Large documents
- Company information
- System instructions
- Knowledge base content

### Dynamic content

Information that changes between requests.

Examples:

- User question
- Current date
- User-specific information

### Recommended structure

```text
STATIC CONTENT
STATIC CONTENT
STATIC CONTENT
----------------
DYNAMIC CONTENT
USER QUESTION
```

### Why?

Keeping the repeated/stable content at the beginning can make it easier for prefix-based caching to reuse that content.

### Memory Trick

> **Stable content first → changing content later**

---

# 🔹 11. Provider-Specific Caching

Different LLM providers can implement prompt caching differently.

For example:

- Some providers support automatic/implicit caching.
- Some allow explicit cache control.
- Pricing for cached tokens can differ.
- Cache duration and behavior can differ.

### Key Idea

> Prompt caching is a general concept, but its implementation and pricing depend on the provider.
