### 🧠 NEW CONCEPTS

### 🔹 1. One-Shot Prompting

#### Definition:
Providing **one example** in the prompt to guide model output.

#### Why:
- Helps model understand format
- Improves structured output

#### Example:
Convert output like this:
```json
{
  "name": "John",
  "age": 20
}
```

👉 Model follows same structure

---

### 🔹 2. Few-Shot (Multi-Shot) Prompting

#### Definition:
Providing **multiple examples** to guide model behavior.

#### Why:
- Improves accuracy
- Useful for complex tasks

#### Example:
```
Input: Apple → Fruit  
Input: Car → Vehicle  
Input: Mango → ?
```

👉 Model learns pattern

---

### 🔹 3. JSON (Structured Output)

#### Definition:
A structured format for organizing data.

#### Example:
```json
{
  "name": "Pravalika",
  "skills": ["Python", "AI"]
}
```

#### Why:
- Easy to read
- Easy to use in code
- Organized output

---

### 🔹 4. Forcing JSON Output

#### Definition:
Restricting model to return output ONLY in JSON format.

#### Code:
```python
response_format = {"type": "json_object"}
```

#### Why:
- Avoids messy text
- Ensures consistent structure

---

### 🔹 5. json.loads()

#### Definition:
Converts JSON string → Python dictionary

#### Example:

Before:
```python
"{ \"name\": \"John\" }"   # string
```

After:
```python
{"name": "John"}   # dictionary
```

#### Why:
- Makes output usable in code

---

### 🔥 COMPLETE FLOW

```
Prompt → LLM → JSON string → json.loads() → Python dictionary
```
