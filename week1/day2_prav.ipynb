```python
import os 
from openai import OpenAI
```

```python
OLLAMA_BASE_URL="http://localhost:11434/v1"
```

```python
messages=[
    {
        "role" : "user",
        "content" : "Tell me a joke"
    }
]
```

```python
print(messages)
```

```python
client = OpenAI(base_url=OLLAMA_BASE_URL, api_key="anything")
response =client.chat.completions.create(model="llama3.2", messages=messages)
print(response.choices[0].message.content)
```

```python
client = OpenAI(base_url=OLLAMA_BASE_URL, api_key="anything")
response =client.chat.completions.create(model="deepscaler:1.5b", messages=messages)
print(response.choices[0].message.content)
```

```python
client = OpenAI(base_url=OLLAMA_BASE_URL, api_key="anything")
response =client.chat.completions.create(model="gemma2:2b", messages=messages)
print(response.choices[0].message.content)
```
