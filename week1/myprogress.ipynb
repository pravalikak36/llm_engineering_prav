### 📅 Day 4 – LLM Engineering Journey

Date: 3 Aug 2026

##### 📚 What I Learned
- Understood how **Transformers** became the foundation of modern LLMs and why the 2017 paper *Attention Is All You Need* was a breakthrough.
- Learned the evolution from **traditional Machine Learning → Neural Networks → Deep Neural Networks → LSTMs → Transformers**.
- Understood **Neural Networks**, **Attention**, and why Transformers are faster due to **parallel processing**.
- Learned the concept of **Emergent Intelligence**—how predicting the next token surprisingly leads to intelligent responses.
- Explored modern AI concepts like **Prompt Engineering**, **Context Engineering**, **Copilots**, and **AI Agents**.
- Learned what **Parameters** are and how they represent the knowledge stored inside a model.
- Understood the difference between:
  - **Training-Time Scaling** (improving the model while training)
  - **Inference-Time Scaling** (making the model perform better while using it)
- Learned why models come in different sizes (Nano, Mini, Full, Haiku, Sonnet, Opus, etc.) and how parameter count affects capability.
- Understood **Tokens**:
  - Why LLMs use tokens instead of characters or full words.
  - Common words vs rare words tokenization.
  - Token rules of thumb (1000 tokens ≈ 750 words).
  - Why code generally consumes more tokens than plain English.
- Explored the GPT Tokenizer and understood how text is converted into tokens before entering an LLM.
