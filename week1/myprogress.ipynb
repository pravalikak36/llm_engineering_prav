---

**📚 Concepts Learned**
- System Prompt
- User Prompt
- Message format (`role` & `content`)
- Why prompts are defined separately
- Why functions improve reusability and maintainability
- How website content is fetched and passed to an LLM

---

**💻 What I Built**
- Built my first Website Summarizer using Ollama.
- Tested the summarizer on different websites.

---

**🧠 Biggest Learnings***
- LLMs work better with well-structured prompts.
- Separating system and user prompts makes code easier to maintain.
- Functions help make code reusable.
- AI output depends on the quality of the input data.
- Many modern job websites are JavaScript-based, so simple web scraping cannot extract complete job descriptions.
