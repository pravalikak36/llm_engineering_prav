```python
import os
from dotenv import load_dotenv
from scraper import fetch_website_contents
from IPython.display import Markdown, display
from openai import OpenAI
```

```python
OLLAMA_BASE_URL = "http://localhost:11434/v1"
```

```python
jobs=fetch_website_contents("https://remoteok.com/remote-ai-jobs")
print(jobs)
```

```python
system_prompt="""
You are a career assistant. Summarize job descriptions and extract the most important information."""
```

```python
user_prompt="""
If this webpage contains a single job posting, extract the required skills, preferred skills, responsibilities, qualifications, and application details. If it contains multiple job listings, summarize the available roles and the common skills they require.
"""
```

```python
def messages_for(website):
    return[
    {
        "role" : "system","content" : system_prompt,
        "role" : "user","content" : user_prompt + website
    }
]
```

```python
messages_for(jobs)
```

```python
def summarize(url):
    website=fetch_website_contents(url)
    client = OpenAI(base_url=OLLAMA_BASE_URL, api_key="anything")
    response=client.chat.completions.create(
        model="llama3.2",
        messages=messages_for(website)
    )
    return response.choices[0].message.content
```

```python
summarize("https://remoteok.com/remote-ai-jobs")
```

```python
def display_summary(url):
    summary = summarize(url)
    display(Markdown(summary))
```

```python
display_summary("https://remoteok.com/remote-ai-jobs")
```
