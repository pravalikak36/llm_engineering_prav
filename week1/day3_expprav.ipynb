## Section 1 :Base, Chat, and Reasoning Models: Understanding LLM Types

### Types of LLMs

#### 1. Base Model
- Predicts the next token/word.
- Trained for text completion.
- Doesn't naturally chat or follow instructions.

#### 2. Chat (Instruct) Model
- Fine-tuned to follow instructions and conversations.
- Uses:
  - System Prompt
  - User Prompt
  - Assistant Response
- Best for general Q&A and chatting.

#### 3. Reasoning Model
- Thinks step by step before answering.
- Better for:
  - Coding
  - Math
  - Logic
  - Complex problem solving

#### 4. Hybrid Model
- Combines Chat + Reasoning.
- Automatically decides how much reasoning is needed.
- Example: GPT-5

---

### Evolution of LLMs

```text
Base Model
     ↓
Chat Model
     ↓
Reasoning Model
     ↓
Hybrid Model
```

---

### Reasoning Budget
- Amount of thinking a reasoning model performs before answering.
- More budget → Better reasoning (usually) but slower response.

### Budget Forcing
- Technique to make a reasoning model spend more time thinking before answering.

---

### Chat vs Reasoning

| Chat Model | Reasoning Model |
|------------|-----------------|
| Faster | Slower |
| General conversations | Complex reasoning |
| Lower cost | Higher cost |
| Q&A, writing | Coding, math, logic |

## Section 2 : Frontier LLMs

### Strengths
- Excellent at information synthesis & summarization.
- Generate high-quality content (emails, reports, presentations, etc.).
- Strong coding assistance (generate, debug, explain code).

---

### Limitations

#### 1. Knowledge Cutoff
- Models only know data from their training period.
- Recent information requires external tools (e.g., web search).

#### 2. Hallucinations
- LLMs predict the most likely next token.
- Can generate incorrect answers with high confidence.

#### 3. Human Supervision
- Best used as an assistant, not an authority.
- Always verify AI-generated code and important information.

---

### Key Takeaway
Think of an LLM as a **junior developer**:
- Fast
- Helpful
- Productive
- But still needs human review.
