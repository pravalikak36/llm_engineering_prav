## Section 1 : Rise of Transformers

### GPT (Generative Pre-trained Transformer)

#### GPT
- **G** → Generative (Generates/Predicts next token)
- **P** → Pre-trained (Trained on huge amounts of data)
- **T** → Transformer (Architecture used to build modern LLMs)

---

### Neural Network

- ML model loosely inspired by the human brain.
- Consists of many interconnected **artificial neurons**.
- Each neuron learns a small pattern.
- Together they learn complex patterns and make predictions.

### Deep Neural Network
- Neural Network with many layers.
- More layers → Learns more complex patterns.
- Also called **Deep Learning**.

---

### Transformer

- Introduced by Google's 2017 paper:
  **"Attention Is All You Need"**
- A **Transformer** is a special architecture of a Deep Neural Network.
- Designed to handle sequences (like text) efficiently.

##### Why Transformer?
- Uses **Attention** to focus on the most important words/tokens.
- Faster training.
- Better scalability.
- Handles larger datasets and models efficiently.

---

### Evolution

```text
Traditional ML
      ↓
Neural Network
      ↓
Deep Neural Network
      ↓
Transformer (2017)
      ↓
GPT Models
      ↓
ChatGPT
```

---

### GPT Timeline

- **2018** → GPT-1
- **2019** → GPT-2
- **2020** → GPT-3
- **2022** → ChatGPT (GPT-3.5 + RLHF)
- **2023** → GPT-4
- **2024** → GPT-4o
- **Current** → GPT-5

---

### ChatGPT

- Built on GPT.
- Uses **RLHF (Reinforcement Learning from Human Feedback)**.
- Learns conversations using:
  - System Prompt
  - User Prompt
  - Assistant Response

---

### Key Takeaways

- GPT predicts the **next token**.
- Transformer is the architecture behind GPT.
- Attention helps the model focus on important information.
- Transformer made LLMs faster, cheaper, and more scalable.
- Transformer is currently the dominant architecture, but alternatives are being researched.

## Section 2 : Transformer,Prompt Engineering,Context Engineering,Agentic AI introduction

### Before Transformers

#### LSTM (Long Short-Term Memory)
- Type of Recurrent Neural Network (RNN).
- Good at understanding sequences.
- Processes data **step by step (sequentially)**.
- Difficult to parallelize → Slow training.

---

### Transformer

- Introduced by Google's 2017 paper:
  **"Attention Is All You Need"**
- Simplified the previous architecture.
- Uses **Attention** to focus on important information.
- Can process data **in parallel**.
- Faster training, better scalability, lower cost.

---

### Emergent Intelligence

- LLMs are trained to predict the next token.
- As models become larger and train on more data, new abilities emerge.
- Examples:
  - Reasoning
  - Coding
  - Math
  - Question Answering

> Large models begin to imitate intelligent behavior.

---

### Prompt Engineering

- Writing effective prompts to get better responses.
- Includes:
  - Giving context
  - Defining a role
  - Specifying output format
  - Providing examples

---

### Context Engineering

- Extension of Prompt Engineering.
- Focuses on providing the **right information** to the LLM.
- Goal: Give the model everything it needs to produce accurate responses.

---

### Agentic AI

- LLM controls the workflow.
- Can:
  - Plan tasks
  - Use tools
  - Call other LLMs
  - Repeat actions until the goal is completed.

##### Agent Flow

```text
User
   ↓
LLM
   ↓
Plan
   ↓
Use Tools
   ↓
Think
   ↓
Repeat
   ↓
Final Response
```

---

### Autonomy

- Ability of an LLM to decide its own next action.
- Instead of only answering, it plans and executes tasks autonomously.

---

### Evolution

```text
LSTM
    ↓
Transformer
    ↓
Emergent Intelligence
    ↓
Prompt Engineering
    ↓
Context Engineering
    ↓
Agentic AI
```

## Section 3 : Parameters,Inference Time Scaling,Training Time scaling,MOE,Chinchilla Scalling laws Introduction

### Parameters

- Parameters are the **learned values (weights)** inside a neural network.
- They store the knowledge learned during training.
- Generally:
  - More parameters → More learning capacity → Smarter model.

###### Examples

- GPT-1 → 117M
- GPT-2 → 1.5B
- GPT-3 → 175B
- GPT-4 → ~1.76T
- Latest Frontier Models → Probably Tens of Trillions

---

### Logarithmic Scale (Log Scale)

The graph uses a **log scale**, not a linear scale.

Linear Scale:

1 → 2 → 3 → 4

Log Scale:

1B → 10B → 100B → 1T → 10T

> Each step is **10× larger** than the previous one.

---

### Training Time

Training = Teaching the model.

During training:
- Model learns from huge datasets.
- Parameters are updated.
- Knowledge is stored inside the model.

Flow:

Internet + Books + Code
        ↓
 Train the Model
        ↓
 Update Parameters
        ↓
 Finished LLM

---

### Training Time Scaling

Improving the model **while training**.

Includes:
- More parameters
- More training data
- More compute (GPUs)

Result:

Bigger Model
      ↓
Learns More
      ↓
Better Performance

---

### Inference

Inference = Using an already trained model.

User asks a question.
The model does **not learn**.
It simply generates an answer using its learned knowledge.

Flow:

User Prompt
      ↓
Already Trained Model
      ↓
Response

---

### Inference Time Scaling

Improving answers **without retraining** the model.

Techniques:
- Better prompts
- Chain of Thought / Reasoning
- More context
- RAG
- Tools

Flow:

Same Model
      ↓
Better Prompt / Context
      ↓
Better Response

---

### Training Time vs Inference Time

Training Time
- Model learns.
- Parameters change.
- Expensive.
- Happens once before release.

Inference Time
- Model answers questions.
- Parameters do NOT change.
- Fast.
- Happens whenever users interact.

---

### Chinchilla Scaling Law

General idea:

More Parameters
        ↓
Can effectively learn from
more training data.

---

### Mixture of Experts (MoE)

Instead of one giant model working every time,

Many smaller expert models exist inside.

Question
      ↓
Relevant Expert Activated
      ↓
Generate Answer

Benefits:
- Faster
- More efficient
- Better performance

---

### Model Variants

Same model released in different sizes.

Examples:

GPT-5
• Nano
• Mini
• Full

Claude
• Haiku
• Sonnet
• Opus

Smaller Model
✔ Faster
✔ Cheaper

Larger Model
✔ More capable
✔ More expensive

---

### Key Takeaway

There are **two ways to improve LLM performance**:

1. **Training Time Scaling**
   - Better model by increasing parameters, data, and compute.

2. **Inference Time Scaling**
   - Better answers using reasoning, prompts, context, RAG, and tools without changing the model.

## Section 4 : Introduction to Tokens

### Tokens

A Token is a **chunk of text** used as the input to an LLM.

A token can be:
- A character
- Part of a word
- A whole word
- Multiple common words

---

### Evolution

##### 1. Character Level

Example:

L | e | a | r | n

Advantages:
- Small vocabulary

Disadvantages:
- Model has to learn both spelling and meaning.
- Hard to train.

---

##### 2. Word Level

Example:

Learning | AI | is | fun

Advantages:
- Easier to understand words.

Disadvantages:
- Huge vocabulary.
- Rare/new words are difficult to handle.

---

##### 3. Token Level (Current Approach)

Example:

Learn | ing | AI | is | fun

Advantages:
- Smaller vocabulary.
- Easier to train.
- Handles new and rare words efficiently.
- Learns common word patterns.

---

### Why Tokens?

Tokens provide a balance between:

- Character-level (too small)
- Word-level (too large)

They make training faster and more efficient.

---

### Token ID

Each token is assigned a unique number.

Example:

"Hello"
      ↓
Token ID

The neural network receives **Token IDs** as input.

---

### Token vs Vector

Token
- Chunk of text.

Vector
- Numerical representation created inside the neural network.

Flow:

Text
   ↓
Tokenizer
   ↓
Token IDs
   ↓
Vectors
   ↓
LLM Processing

---

### Key Takeaway

Characters → Too difficult

Words → Too many possibilities

Tokens → Best balance between efficiency and understanding.

## Section 5 : Tokenization Rules and Examples

### Common Words

Common words usually become **one token**.

Example:

I | am | learning | AI

---

### Tokens Include Spaces

Tokens often include the **space before a word**.

Example:

" learning"

instead of

"learning"

This tells the model that a new word begins.

---

### Rare Words

Rare words are split into multiple tokens.

Example:

handcrafted

↓

hand | crafted

Example:

exquisitely

↓

ex | quis | ite | ly

This helps the model understand unfamiliar words.

---

### Numbers

Large numbers are also split into tokens.

Example:

314159

↓

314 | 159

Older models sometimes struggled with larger numbers because they were split across multiple tokens.

---

### Rule of Thumb

- 1 Token ≈ 4 Characters
- 1000 Tokens ≈ 750 Words

---

# Code vs English

English text usually uses fewer tokens.

Code uses more tokens because of:
- Variable names
- Symbols
- Keywords
- Punctuation

---

### GPT Tokenizer

Used to:
- View tokenization.
- Count tokens.
- See Token IDs.

Useful for estimating context size and API cost.
