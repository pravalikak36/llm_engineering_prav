## Section 1 : Rise of Transformers

### GPT (Generative Pre-trained Transformer)

#### GPT
- **G** → Generative (Generates/Predicts next token)
- **P** → Pre-trained (Trained on huge amounts of data)
- **T** → Transformer (Architecture used to build modern LLMs)

---

### Neural Network

- ML model loosely inspired by the human brain.
- Consists of many interconnected **artificial neurons**.
- Each neuron learns a small pattern.
- Together they learn complex patterns and make predictions.

### Deep Neural Network
- Neural Network with many layers.
- More layers → Learns more complex patterns.
- Also called **Deep Learning**.

---

### Transformer

- Introduced by Google's 2017 paper:
  **"Attention Is All You Need"**
- A **Transformer** is a special architecture of a Deep Neural Network.
- Designed to handle sequences (like text) efficiently.

##### Why Transformer?
- Uses **Attention** to focus on the most important words/tokens.
- Faster training.
- Better scalability.
- Handles larger datasets and models efficiently.

---

### Evolution

```text
Traditional ML
      ↓
Neural Network
      ↓
Deep Neural Network
      ↓
Transformer (2017)
      ↓
GPT Models
      ↓
ChatGPT
```

---

### GPT Timeline

- **2018** → GPT-1
- **2019** → GPT-2
- **2020** → GPT-3
- **2022** → ChatGPT (GPT-3.5 + RLHF)
- **2023** → GPT-4
- **2024** → GPT-4o
- **Current** → GPT-5

---

### ChatGPT

- Built on GPT.
- Uses **RLHF (Reinforcement Learning from Human Feedback)**.
- Learns conversations using:
  - System Prompt
  - User Prompt
  - Assistant Response

---

### Key Takeaways

- GPT predicts the **next token**.
- Transformer is the architecture behind GPT.
- Attention helps the model focus on important information.
- Transformer made LLMs faster, cheaper, and more scalable.
- Transformer is currently the dominant architecture, but alternatives are being researched.
