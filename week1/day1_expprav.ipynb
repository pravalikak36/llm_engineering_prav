### 📘 Section 1 - Imports

**What is an import?**

An import is a way of bringing an existing Python module (library) into our program so that we can use its functions, classes, and other features without writing them ourselves.

---
**Why do we use imports?**

Python doesn't know everything by default. Different modules provide different capabilities.

Instead of writing all the code ourselves, we import the module we need and use the functions it already provides.

For example:

* math → Mathematical operations
* random → Generate random numbers
* os → Interact with the operating system
* numpy → Numerical computations
* pandas → Data analysis
---
**What is a module?**

A module is a Python file that contains related functions, classes, and variables.

Think of it like a toolbox. Each toolbox is designed for a specific purpose.

For example:

* The math module contains mathematical tools.
* The random module contains random number generators.
* The os module contains operating system utilities.
---
**Types of imports**
1. Import the entire module


```python
import math
```

```python
math.sqrt(16)
```

2. Import only what you need  

```python
from math import sqrt  
sqrt(16) 
```

3. Import with an alias

```python
import math as m 
math.sqrt(16)
m.sqrt(16) 
```

### 📘 Section 2 – Some Conditions 
* What does not mean?

not reverses a condition.

Example:

```python
logged_in = True
if not logged_in:
    print("Please log in")

# This will print "Please log in" because logged_in is True
```

 * **startswith()**

```python
name="John Doe "
if name.startswith("John"):
    print("Hello, John!")

# This will print "Hello, John!" because name starts with "John"

else:
    print("Hello, stranger!")
```

* **strip()**

```python
Password = "  1234567890   "
Password.strip()
# This will remove the spaces from the beginning and end of the string
```

* **if-else condition**

```python
age=20
if age>18:
    print("You are eligible to vote")
elif age>=18:
    print("You are eligible to vote")
else:
    print("You are not eligible to vote")

```

### 📘Section 3 : Messages
**What are messages?**

Messages are the format used to communicate with an LLM. They store the conversation between the user and the AI

---
**Structure**

```python
messages=[
    {
        "role" : "user",
        "content" : "Hey!How are you??"
    }
]
```

**Why messages (plural)?**

Because a conversation contains multiple messages, not just one.

---

**[] (List)**

Stores multiple messages in a conversation.

---

**{} (Dictionary)**

Stores information about a single message as key-value pairs.

---

**role**

Specifies who is sending the message.

---

**Common roles:**

system → Instructions for the AI.
user → User's prompt.
assistant → AI's previous response.

----

**content**

The actual text of the message.


```python
messages=[
    {
        "role" : "system",
        "content" : "Act as a techer proficient in programming skills",
        "role" : "user",
        "content" : "Explain Python"
        
     }
]
```

---
**Works for OpenAI & Ollama**

The message format is essentially the same for both chat-based APIs.

### 📘Section 4 : Connecting OpenAI SDK with Ollama
**Goal**

Instead of sending requests to OpenAI's cloud servers, we configure the OpenAI SDK to send requests to our local Ollama server.

---

```python
from openai import OpenAI

OLLAMA_BASE_URL = "http://localhost:11434/v1"

client = OpenAI(
    base_url=OLLAMA_BASE_URL,
    api_key="anything"
)

response = client.chat.completions.create(
    model="llama3.2",
    messages=[
        {
            "role": "user",
            "content": "Hello!"
        }
    ]
)

print(response.choices[0].message.content)
```

#### Understanding Each Line

1.

```python
from openai import OpenAI
```

Imports the OpenAI Python SDK.

Even though we're using Ollama, we still use the OpenAI SDK because Ollama supports the OpenAI API format.

---

2.

```python
OLLAMA_BASE_URL = "http://localhost:11434/v1"
```

Specifies where requests should be sent.

* localhost → Your own computer.
* 11434 → Default Ollama server port.
* /v1 → OpenAI-compatible API endpoint.

---

3.

```python
client = OpenAI(...)
```

Creates a client object used to communicate with the LLM.

Normally it communicates with OpenAI.

By changing base_url, it communicates with Ollama instead.

---

4.

```python
api_key="anything"
```

The OpenAI SDK requires an API key parameter.

Ollama ignores it, so any string works.

---

5.

```python
client.chat.completions.create(...)
```

Sends a chat request to the model.

This is where the actual conversation happens.

---

6.

```python
model="llama3.2"
```

Specifies which local model should answer.

Examples:

* llama3.2
* gemma3
* qwen2.5

---

7.

```python
messages=[...]
```

Contains the conversation history.

Each message has:

* role
* content

---

8.

```python
print(response.choices[0].message.content)
```

Extracts the model's reply from the response object.

**Q: Why use the OpenAI SDK with Ollama instead of the Ollama Python library?**

Answer:

It provides a standard API compatible with OpenAI.
The same application can easily switch between local and cloud models by changing the configuration.
Many LLM providers support the OpenAI-compatible API, making the code more portable.
