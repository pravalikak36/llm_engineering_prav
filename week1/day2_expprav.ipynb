# Section 1-Frontier and Open Source Models

### Frontier Models
- Most advanced LLMs built by major AI companies (Frontier Labs).
- Usually refers to **closed-source** models.

#### Closed-Source Models
- Paid to use via APIs.
- Examples:
  - GPT (OpenAI)
  - Claude (Anthropic)
  - Gemini (Google)
  - Grok (xAI)

#### Open-Source (Open-Weight) Models
- Can be downloaded & run locally.
- Companies release **model weights**, but not always the training data.
- Examples:
  - Llama (Meta)
  - Mistral
  - Qwen
  - Gemma
  - Phi
  - DeepSeek
  - GPT-OSS

---

### 3 Ways to Use LLMs

#### 1. Products (UI)
- Ready-to-use AI applications.
- Examples: ChatGPT, Claude, Gemini.

#### 2. Cloud APIs
- Developers integrate LLMs into applications using API calls.
- Examples: OpenAI API, Azure OpenAI, Vertex AI, Bedrock, OpenRouter.

#### 3. Run Locally
- Download and run open-source models on your own computer.
- Example: Ollama + Llama/Gemma/Phi.

---

### Ollama

- Runs open-source LLMs **locally**.
- Install Ollama software.
- Download a model (`ollama pull llama3.2`).
- Python sends request:
  ```python
  client.chat.completions.create(...)
  ```
- Ollama performs inference locally.
- Response is returned to your program.

### Flow
```
Python
    ↓
client.chat.completions.create()
    ↓
Ollama
    ↓
Local LLM
    ↓
Response
```

---

### Hugging Face

- Platform to discover and use AI models.
- Can:
  - Download models and run them locally.
  - Use models through Hugging Face APIs.

### Ollama vs Hugging Face

| Ollama | Hugging Face |
|--------|--------------|
| Runs models locally | Models can run locally or via cloud API |
| Focused on local inference | Model hub + APIs + datasets + tools |

---

### Key Takeaways
- **Frontier Models** → Most advanced LLMs.
- **Closed-source** → Paid APIs.
- **Open-source/Open-weight** → Can run locally.
- **LLMs can be used via:** Products, APIs, or Local execution.
- **Ollama** → Local model runner.
- **Hugging Face** → Platform for models, datasets & AI tools.
